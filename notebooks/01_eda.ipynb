```python
DATA_PATH = Path("../data/raw/UCI_Credit_Card.csv")
REPORT_DIR = Path("../reports/eda")
REPORT_DIR.mkdir(parents=True, exist_ok=True)
```

```python
df = pd.read_csv(DATA_PATH)

print("Размер датасета:", df.shape)
display(df.head())
```

```python

print("Информация о данных:")
df.info()

print("\nПропущенные значения:")
display(df.isnull().sum().to_frame("missing_values"))

print("\nДубликаты строк:", df.duplicated().sum())
print("Дубликаты ID:", df["ID"].duplicated().sum())

print("\nОписательная статистика:")
display(df.describe().T)
```

```python

target = "default.payment.next.month"

counts = df[target].value_counts().sort_index()

plt.figure(figsize=(8, 5))

ax = sns.barplot(
    x=["Без дефолта", "Дефолт"],
    y=counts.values,
)

for i, count in enumerate(counts.values):
    ax.text(
        i,
        count + 200,
        f"{count:,} ({count / len(df):.1%})",
        ha="center",
    )

plt.title("Распределение целевой переменной")
plt.ylabel("Количество клиентов")
plt.xlabel("")
plt.tight_layout()

plt.savefig(
    REPORT_DIR / "target_distribution.png",
    dpi=150,
)
plt.show()
```

```python

plt.figure(figsize=(10, 5))

sns.histplot(
    data=df,
    x="AGE",
    hue=target,
    bins=30,
    multiple="stack",
)

plt.title("Распределение возраста клиентов")
plt.xlabel("Возраст")
plt.ylabel("Количество клиентов")
plt.tight_layout()

plt.savefig(
    REPORT_DIR / "age_distribution.png",
    dpi=150,
)
plt.show()
```

```python

df["AGE_GROUP"] = pd.cut(
    df["AGE"],
    bins=[17, 25, 35, 45, 60, 100],
    labels=["18-25", "26-35", "36-45", "46-60", "61+"],
)

age_default = (
    df.groupby("AGE_GROUP", observed=True)[target]
    .agg(["mean", "count"])
    .reset_index()
)

display(age_default)

plt.figure(figsize=(9, 5))

sns.barplot(
    data=age_default,
    x="AGE_GROUP",
    y="mean",
)

plt.title("Доля дефолтов по возрастным группам")
plt.xlabel("Возрастная группа")
plt.ylabel("Доля дефолтов")

plt.tight_layout()

plt.savefig(
    REPORT_DIR / "default_by_age.png",
    dpi=150,
)
plt.show()
```

```python

plt.figure(figsize=(10, 5))

sns.boxplot(
    data=df,
    x=target,
    y="LIMIT_BAL",
    showfliers=False,
)

plt.title("Кредитный лимит и дефолт")
plt.xlabel("Дефолт в следующем месяце")
plt.ylabel("Кредитный лимит")

plt.tight_layout()

plt.savefig(
    REPORT_DIR / "credit_limit_boxplot.png",
    dpi=150,
)
plt.show()
```

```python

selected_columns = [
    "LIMIT_BAL",
    "AGE",
    "PAY_0",
    "PAY_2",
    "BILL_AMT1",
    "BILL_AMT2",
    "PAY_AMT1",
    "PAY_AMT2",
    target,
]

corr = df[selected_columns].corr()

plt.figure(figsize=(11, 8))

sns.heatmap(
    corr,
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    center=0,
)

plt.title("Корреляционная матрица")
plt.tight_layout()

plt.savefig(
    REPORT_DIR / "correlation_matrix.png",
    dpi=150,
)
plt.show()
```

4. Проанализированы распределения возраста, кредитного лимита и дефолтов по возрастным группам.
5. Построена корреляционная матрица ключевых финансовых признаков.

**Вывод:** для построения PD-модели необходимо учитывать дисбаланс классов, историю просрочек и финансовые характеристики клиентов. При оценке модели используются ROC-AUC, Precision, Recall и F1-Score.
